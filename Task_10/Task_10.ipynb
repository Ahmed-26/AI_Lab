```python
import pandas as pd
```

```python
class DataPreProcessor:
    def __init__(self, file_path):
        self.file_path = file_path
        self.data = None
    
    def load_data(self):
        self.data = pd.read_csv(self.file_path)
        return self.data
    
    def get_shape(self):
        return self.data.shape
    
    def get_info(self):
        return self.data.info()
    
    def get_description(self):
        return self.data.describe()
```

```python
class NullValueHandler:
    def __init__(self, data):
        self.data = data
    
    def get_null_summary(self):
        return self.data.isnull().sum()
```

```python
class DataTypeConverter:
    def __init__(self, data):
        self.data = data
    
    def get_dtypes(self):
        return self.data.dtypes
    
    def convert_to_int(self, column):
        self.data[column] = self.data[column].astype(int)
        return self.data
    
    def convert_columns_to_int(self, columns):
        for col in columns:
            self.data[col] = self.data[col].astype(int)
        return self.data
    
    def convert_all_numeric_to_int_except(self, exclude_columns):
        numeric_cols = self.data.select_dtypes(include=['float64']).columns
        for col in numeric_cols:
            if col not in exclude_columns:
                self.data[col] = self.data[col].astype(int)
        return self.data
```

```python
# Instantiate all objects and demonstrate usage
preprocessor = DataPreProcessor('diabetes.csv')
df = preprocessor.load_data()
print("Data Shape:", preprocessor.get_shape())
print("\nData Description:")
print(preprocessor.get_description())
print("\nFirst 5 rows:")
print(df.head())

null_handler = NullValueHandler(df)
print("\nNull Values Summary:")
print(null_handler.get_null_summary())

converter = DataTypeConverter(df)
print("\nData Types Before Conversion:")
print(converter.get_dtypes())

df = converter.convert_all_numeric_to_int_except(['Screen Size (inches)'])
print("\nData Types After Conversion:")
print(converter.get_dtypes())

print("\nFirst 5 rows after conversion:")
print(df.head())
```
