```python
import pandas as pd
```

```python
class Load_data:
    def __init__(self, file_path):
        self.file_path = file_path
        self.data = None
    
    def load_csv(self):
        self.data = pd.read_csv(self.file_path)
        return self.data
    
    def get_data(self):
        return self.data

```

```python
class DataExplorer:
    def __init__(self, data):
        self.data = data
    
    def get_shape(self):
        return self.data.shape
    
    def get_info(self):
        return self.data.info()
    
    def get_description(self):
        return self.data.describe()
    
    def get_null_counts(self):
        return self.data.isnull().sum()
    
    def get_columns(self):
        return self.data.columns.tolist()
    
    def get_dtypes(self):
        return self.data.dtypes
    
```

```python

loader = Load_data('diabetes.csv')
df = loader.load_csv()
print("First 5 rows of the dataset:")
print(df.head())
print("\n" + "="*50 + "\n")

explorer = DataExplorer(df)
print("Shape:", explorer.get_shape())
print("\nColumns:", explorer.get_columns())
print("\nData Types:")
print(explorer.get_dtypes())
print("\nNull Counts:")
print(explorer.get_null_counts())
print("\nStatistical Description:")
print(explorer.get_description())
```
